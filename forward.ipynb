```
import numpy as np
import torch
import pandas as pd
import matplotlib.pyplot as plt
```

```
# Set random seed for reproducible results
torch.manual_seed(42)
# ==========================================
# 1. DEFINE ARCHITECTURE & PARAMETERS
# ==========================================
input_dim = 10      # Number of input features
hidden1_dim = 16    # Hidden Layer 1 neurons
hidden2_dim = 16     # Hidden Layer 2 neurons
hidden3_dim =8      # Hidden Layer 2 neurons
output_dim = 4      # Output classes

# Layer 1 Weights and Biases (Input -> Hidden 1) 10x16
W1 = torch.randint(low=0, high=10, size=(input_dim, hidden1_dim)).float()
b1 = torch.ones(1, hidden1_dim)

# Layer 2 Weights and Biases (Hidden 1 -> Hidden 2) 16x16
W2 = torch.randn(hidden1_dim, hidden2_dim)
b2 = torch.ones(1, hidden2_dim)
# Layer 3 Weights and Biases (Hidden 2 -> Hidden 3) 16x8
W3 = torch.randn(hidden2_dim, hidden3_dim)
b3 = torch.ones(1, hidden3_dim)
# Layer 3 Weights and Biases (Hidden 3 -> Output) 8x4
W4 = torch.randn(hidden3_dim, output_dim)
b4 = torch.ones(1, output_dim)
#print(W1.shape ,  W2.shape,  W3.shape, W4.shape, )

# ==========================================
# 2. DEFINE ACTIVATION FUNCTIONS
# ==========================================
def sigmoid(x):
    return 1.0 / (1.0 + torch.exp(-x))

def relu(x):
    return torch.maximum(torch.tensor(0.0), x)

def softmax(x):
    # Subtract max per row for numerical stability
    exp_x = torch.exp(x - torch.max(x, dim=1, keepdim=True).values)
    return exp_x / torch.sum(exp_x, dim=1, keepdim=True)


# ==========================================
# 3. FORWARD PASS
# ==========================================
# Dummy input data: Batch of 4 samples, 10 features each
X = torch.randn(4, input_dim)

# Layer 1: Linear transformation + Sigmoid activation
z1 = torch.matmul(X, W1) + b1
a1 = sigmoid(z1)

# Layer 2: Linear transformation + ReLU activation
z2 = torch.matmul(a1, W2) + b2
a2 = relu(z2)
#print(a2.shape,W3.shape)
#Layer 3:
z3 = torch.matmul(a2, W3) + b3
a3 = sigmoid(z3)
# Output Layer: Linear transformation + Softmax activation
z4 = torch.matmul(a3, W4) + b4
output_probabilities = softmax(z4)


# ==========================================
# 4. RESULTS
# ==========================================
print("Output Probabilities (Shape:", output_probabilities.shape, "):")
print(output_probabilities)

print("\nCheck row sums (should all be 1.0):")
print(output_probabilities.sum(dim=1))
```
